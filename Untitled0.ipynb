```python
# ==========================================================
# CHIA LẠI DATASET theo identity-disjoint split (chuẩn FaceForensics++)
# >>> SCRIPT VERSION: v3 (có rmtree dọn sạch trước khi copy) <<<
#
# Vấn đề: dataset cũ chia theo VIDEO, không theo DANH TÍNH -> 1 người có
# thể vừa có video real ở val/test, vừa "xuất hiện" (qua video fake) ở
# train -> leakage đã phát hiện qua check_identity_leakage.py.
#
# Cách fix: dùng Union-Find gom các identity ID liên kết với nhau qua
# file fake (target_source) thành từng CỤM, rồi chia nguyên cụm (không
# tách lẻ) vào train/val/test -> đảm bảo không còn danh tính nào bị
# rải qua nhiều tập.
# ==========================================================

import os
import re
import shutil
import random
from collections import defaultdict

random.seed(42)

# ----------------------------
# CONFIG
# ----------------------------
OLD_DATASET_ROOT = "/content/output_dataset"          # dataset cũ (bị leak)
NEW_DATASET_ROOT  = "/content/output_dataset_resplit"  # dataset mới, đã chia đúng
SPLIT_RATIOS = {"train": 0.8, "val": 0.1, "test": 0.1}  # tỉ lệ mong muốn (theo số frame)
DRY_RUN = False   # True: chỉ in thống kê, KHÔNG copy file. Đổi False sau khi xem thống kê ổn.

REAL_PATTERN = re.compile(r'original_(\d+)_f\d+\.(jpg|png|jpeg)$', re.IGNORECASE)
FAKE_PATTERN = re.compile(
    r'(Deepfakes|Face2Face|FaceSwap|NeuralTextures|FaceShifter)_(\d+)_(\d+)_f\d+\.(jpg|png|jpeg)$',
    re.IGNORECASE
)

# ----------------------------
# 1. QUÉT TOÀN BỘ FILE từ 3 tập cũ (gộp lại vì split cũ không đáng tin)
# ----------------------------
all_real_files = []   # (path, identity_id)
all_fake_files = []   # (path, id1, id2)
skipped = []

for split in ["train", "val", "test"]:
    for cls in ["real", "fake"]:
        folder = os.path.join(OLD_DATASET_ROOT, split, cls)
        if not os.path.isdir(folder):
            continue
        for fname in os.listdir(folder):
            path = os.path.join(folder, fname)
            if cls == "real":
                m = REAL_PATTERN.match(fname)
                if m:
                    all_real_files.append((path, m.group(1)))
                else:
                    skipped.append(path)
            else:
                m = FAKE_PATTERN.match(fname)
                if m:
                    all_fake_files.append((path, m.group(2), m.group(3)))
                else:
                    skipped.append(path)

print(f"Real files: {len(all_real_files)} | Fake files: {len(all_fake_files)} | "
      f"Bỏ qua (không khớp pattern): {len(skipped)}")
if skipped:
    print("Vài file bị bỏ qua (kiểm tra lại pattern nếu số lượng nhiều):")
    for p in skipped[:5]:
        print(" -", p)

# ----------------------------
# 2. UNION-FIND: gom các identity ID liên kết với nhau qua file fake
# ----------------------------
parent = {}

def find(x):
    parent.setdefault(x, x)
    while parent[x] != x:
        parent[x] = parent[parent[x]]  # path compression
        x = parent[x]
    return x

def union(a, b):
    ra, rb = find(a), find(b)
    if ra != rb:
        parent[ra] = rb

for _, id1, id2 in all_fake_files:
    union(id1, id2)
for _, ident in all_real_files:
    find(ident)  # đảm bảo identity chỉ có real cũng được đăng ký vào union-find

# ----------------------------
# 3. Gom file vào từng CỤM (component) theo root
# ----------------------------
component_real = defaultdict(list)   # root -> list of real file paths
component_fake = defaultdict(list)   # root -> list of fake file paths

for path, ident in all_real_files:
    component_real[find(ident)].append(path)
for path, id1, id2 in all_fake_files:
    component_fake[find(id1)].append(path)   # find(id1) == find(id2), dùng cái nào cũng được

all_roots = set(component_real.keys()) | set(component_fake.keys())
component_sizes = {r: len(component_real[r]) + len(component_fake[r]) for r in all_roots}

print(f"\nTổng số cụm (component) sau khi gom: {len(all_roots)}")
sizes_sorted = sorted(component_sizes.values(), reverse=True)
print(f"Cụm lớn nhất: {sizes_sorted[0]} file | Cụm nhỏ nhất: {sizes_sorted[-1]} file")
print(f"Top 5 cụm lớn nhất (số file): {sizes_sorted[:5]}")
total_files = sum(sizes_sorted)
print(f"Cụm lớn nhất chiếm {sizes_sorted[0] / total_files * 100:.1f}% tổng dataset")

if sizes_sorted[0] / total_files > 0.5:
    print("\n⚠️  CẢNH BÁO: có 1 cụm chiếm hơn 50% dataset — các danh tính bị liên kết "
          "chéo quá nhiều qua nhiều kỹ thuật fake khác nhau, khó chia tách 80/10/10 "
          "cho đẹp. Script vẫn chạy tiếp nhưng tỉ lệ thực tế có thể lệch khá xa mục tiêu.")

total_real = sum(len(v) for v in component_real.values())
total_fake = sum(len(v) for v in component_fake.values())
print(f"\nTỉ lệ real/fake TOÀN BỘ dataset: real={total_real} ({total_real/total_files*100:.1f}%), "
      f"fake={total_fake} ({total_fake/total_files*100:.1f}%)")
print("-> Đây là tỉ lệ mục tiêu mà mỗi tập (train/val/test) cần bám theo, "
      "không chỉ cân bằng tổng số file.")

# ----------------------------
# 4. Chia CỤM (không tách lẻ) vào train/val/test - greedy CÂN BẰNG CẢ 2 CHIỀU:
#    (a) tổng kích thước mỗi tập theo tỉ lệ mục tiêu
#    (b) tỉ lệ real/fake bên trong mỗi tập bám theo tỉ lệ real/fake toàn dataset
# Với mỗi cụm, tính "chi phí" nếu gán vào từng tập = mức lệch tăng thêm so với
# mục tiêu (cả về tổng size lẫn về số lượng real/fake riêng) -> chọn tập có
# chi phí thấp nhất.
# ----------------------------
sorted_roots = sorted(all_roots, key=lambda r: component_sizes[r], reverse=True)

real_ratio = total_real / total_files
fake_ratio = total_fake / total_files

split_totals = {s: 0 for s in SPLIT_RATIOS}
split_real = {s: 0 for s in SPLIT_RATIOS}
split_fake = {s: 0 for s in SPLIT_RATIOS}
split_targets = {s: ratio * total_files for s, ratio in SPLIT_RATIOS.items()}
split_assignment = {}  # root -> split name

for root in sorted_roots:
    n_real = len(component_real[root])
    n_fake = len(component_fake[root])
    n_total = n_real + n_fake

    best_split, best_score = None, -float("inf")
    for s in SPLIT_RATIOS:
        target_total = split_targets[s]
        target_real = target_total * real_ratio
        target_fake = target_total * fake_ratio

        # Độ thiếu hụt TƯƠNG ĐỐI (chia cho mục tiêu) - so sánh công bằng giữa
        # tập có mục tiêu lớn (train) và tập có mục tiêu nhỏ (val/test).
        # Càng "thiếu" nhiều theo tỉ lệ -> score càng cao -> càng ưu tiên gán.
        deficit_total = (target_total - split_totals[s]) / target_total
        deficit_real  = (target_real - split_real[s]) / target_real if target_real > 0 else 0
        deficit_fake  = (target_fake - split_fake[s]) / target_fake if target_fake > 0 else 0

        score = deficit_total + deficit_real + deficit_fake

        if score > best_score:
            best_score = score
            best_split = s

    split_assignment[root] = best_split
    split_totals[best_split] += n_total
    split_real[best_split] += n_real
    split_fake[best_split] += n_fake

print("\n--- Kết quả chia (theo số file) ---")
for s in SPLIT_RATIOS:
    pct = split_totals[s] / total_files * 100
    r_pct = split_real[s] / split_totals[s] * 100 if split_totals[s] else 0
    f_pct = split_fake[s] / split_totals[s] * 100 if split_totals[s] else 0
    print(f"{s}: {split_totals[s]} file ({pct:.1f}%) - mục tiêu {SPLIT_RATIOS[s]*100:.0f}% "
          f"| real={split_real[s]} ({r_pct:.1f}%) fake={split_fake[s]} ({f_pct:.1f}%) "
          f"- mục tiêu real/fake toàn bộ: {real_ratio*100:.1f}%/{fake_ratio*100:.1f}%")

n_identities_per_split = defaultdict(int)
for root, s in split_assignment.items():
    n_identities_per_split[s] += 1
print("\n--- Số cụm danh tính mỗi tập ---")
for s in SPLIT_RATIOS:
    print(f"{s}: {n_identities_per_split[s]} cụm")

# ----------------------------
# 5. COPY file vào cấu trúc thư mục mới (chỉ chạy thật nếu DRY_RUN=False)
# ----------------------------
if DRY_RUN:
    print("\nDRY_RUN=True -> chưa copy file nào. Xem thống kê ở trên, nếu ổn thì đổi "
          "DRY_RUN=False rồi chạy lại để copy thật.")
else:
    print("\nĐang dọn sạch thư mục đích (nếu có từ lần chạy trước) để tránh sót "
          "file cũ gây trùng lặp/leakage ngược...")
    if os.path.isdir(NEW_DATASET_ROOT):
        shutil.rmtree(NEW_DATASET_ROOT)
        print(f"Đã xoá {NEW_DATASET_ROOT} (bản cũ).")

    print("Đang copy file vào cấu trúc mới...")
    for s in SPLIT_RATIOS:
        os.makedirs(os.path.join(NEW_DATASET_ROOT, s, "real"), exist_ok=True)
        os.makedirs(os.path.join(NEW_DATASET_ROOT, s, "fake"), exist_ok=True)

    copied = 0
    for root, s in split_assignment.items():
        for path in component_real[root]:
            dst = os.path.join(NEW_DATASET_ROOT, s, "real", os.path.basename(path))
            shutil.copy2(path, dst)
            copied += 1
        for path in component_fake[root]:
            dst = os.path.join(NEW_DATASET_ROOT, s, "fake", os.path.basename(path))
            shutil.copy2(path, dst)
            copied += 1
        if copied % 5000 < (len(component_real[root]) + len(component_fake[root])):
            print(f"  ... đã copy {copied}/{total_files}")

    print(f"\nHoàn tất! Đã copy {copied} file vào {NEW_DATASET_ROOT}")
    print("Cập nhật DATASET_ROOT trong script train ResNet18 thành:")
    print(f'  DATASET_ROOT = "{NEW_DATASET_ROOT}"')
```

```python
# ResNet18 Deepfake Detection Baseline
from google.colab import drive
drive.mount('/content/drive')
!pip install facenet-pytorch --no-deps -q

import os
import time
import copy
import json
import random
import shutil
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, balanced_accuracy_score,
    matthews_corrcoef, f1_score, precision_score, recall_score
)

# ----------------------------
# 0. SEED - đặt NGAY sau import, TRƯỚC khi tạo model/dataset/dataloader
# ----------------------------
SEED = 42

def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)
print(f"Đã set seed = {SEED}")

# 1. CONFIG
BATCH_SIZE   = 32
NUM_EPOCHS   = 15
WEIGHT_DECAY = 1e-4
NUM_WORKERS  = 2
USE_HIDDEN_LAYER = False
DROPOUT      = 0.5
FREEZE_MODE  = "partial"
BACKBONE_LR  = 1e-5
HEAD_LR      = 1e-3
EARLY_STOP_PATIENCE = 3
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/resnet18_deepfake_best.pth"

STRATEGY_NAME = "baseline"   # >>> ĐÃ SỬA: trước bị nhầm thành "v4_superheavy"
RESULTS_DIR      = "/content/drive/MyDrive/Deepfakeft/results"
CHECKPOINTS_DIR  = "/content/drive/MyDrive/Deepfakeft/checkpoints"
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINTS_DIR, exist_ok=True)

print(f"Device: {DEVICE}")

# ----------------------------
# Chuẩn bị dataset: copy tar về local rồi giải nén, kiểm tra chữ ký tar
# ----------------------------
TAR_FILE_DRIVE = "/content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar"
TAR_FILE_LOCAL = "/content/output_dataset_resplit.tar"
DATASET_ROOT   = "/content/output_dataset_resplit"
required_dirs  = ["train/real", "train/fake", "val/real", "val/fake", "test/real", "test/fake"]
SIGNATURE_FILE = os.path.join(DATASET_ROOT, ".source_signature")

def get_tar_signature(path):
    stat = os.stat(path)
    return f"{stat.st_size}_{int(stat.st_mtime)}"

def dataset_ready():
    dirs_ok = all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs)
    if not dirs_ok:
        return False
    if not os.path.isfile(SIGNATURE_FILE):
        return False
    with open(SIGNATURE_FILE) as f:
        cached_sig = f.read().strip()
    return cached_sig == get_tar_signature(TAR_FILE_DRIVE)

if not dataset_ready():
    if os.path.isdir(DATASET_ROOT):
        print("Phát hiện tar trên Drive đã đổi (hoặc dataset local không rõ nguồn) "
              "-> xoá dataset local cũ để giải nén lại...")
        get_ipython().system('rm -rf "$DATASET_ROOT"')

    src_size = os.path.getsize(TAR_FILE_DRIVE)
    need_copy = (not os.path.isfile(TAR_FILE_LOCAL)) or (os.path.getsize(TAR_FILE_LOCAL) != src_size)
    if need_copy:
        print(f"Đang copy tar về local ({src_size / 1e9:.2f} GB)...")
        get_ipython().system('cp "$TAR_FILE_DRIVE" "$TAR_FILE_LOCAL"')
        assert os.path.getsize(TAR_FILE_LOCAL) == src_size, "Copy tar bị dở dang/lỗi."
        print("Copy xong.")

    print("Đang giải nén...")
    get_ipython().system('tar -xf "$TAR_FILE_LOCAL" -C /content')
    assert all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs), \
        f"Giải nén xong nhưng thiếu thư mục con trong {DATASET_ROOT}."

    with open(SIGNATURE_FILE, "w") as f:
        f.write(get_tar_signature(TAR_FILE_DRIVE))
    print("Dataset sẵn sàng (đã đồng bộ với tar mới nhất trên Drive).")
else:
    print("Dataset đã sẵn sàng và khớp với tar hiện tại trên Drive.")

# 2. TRANSFORMS
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

# 3. DATASETS & DATALOADERS
train_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "train"), transform=train_transforms)
val_ds   = datasets.ImageFolder(os.path.join(DATASET_ROOT, "val"),   transform=eval_transforms)
test_ds  = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"),  transform=eval_transforms)

print("Class mapping:", train_ds.class_to_idx)
print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

# generator + worker_init_fn để shuffle=True cũng deterministic
g = torch.Generator()
g.manual_seed(SEED)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=NUM_WORKERS,
                       pin_memory=True, persistent_workers=(NUM_WORKERS > 0),
                       worker_init_fn=seed_worker, generator=g)
val_dl   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS,
                       pin_memory=True, persistent_workers=(NUM_WORKERS > 0))
test_dl  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS,
                       pin_memory=True, persistent_workers=(NUM_WORKERS > 0))

# 4. MODEL
def create_model(use_hidden_layer=False, dropout=0.5, freeze_mode="partial"):
    model = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)

    if freeze_mode in ("full", "partial"):
        for param in model.parameters():
            param.requires_grad = False

    if freeze_mode == "partial":
        for param in model.layer4.parameters():
            param.requires_grad = True

    in_features = model.fc.in_features
    print(f'Input feature dim: {in_features}')

    if use_hidden_layer:
        model.fc = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(in_features, in_features // 2),
            nn.ReLU(),
            nn.BatchNorm1d(in_features // 2),
            nn.Dropout(dropout),
            nn.Linear(in_features // 2, 2)
        )
    else:
        model.fc = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(in_features, 2)
        )

    n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    n_total = sum(p.numel() for p in model.parameters())
    print(f"Trainable params: {n_trainable:,} / {n_total:,}")

    return model.to(DEVICE)

model = create_model(USE_HIDDEN_LAYER, DROPOUT, FREEZE_MODE)

# 5. LOSS / OPTIMIZER
criterion_counts = torch.bincount(torch.tensor(train_ds.targets))
class_weights = (criterion_counts.sum() / (len(criterion_counts) * criterion_counts)).float().to(DEVICE)
print(f"Class counts (theo class_to_idx {train_ds.class_to_idx}): {criterion_counts.tolist()}")
print(f"Class weights: {class_weights.tolist()}")

criterion = nn.CrossEntropyLoss(weight=class_weights)

backbone_params = [p for n, p in model.named_parameters() if p.requires_grad and not n.startswith("fc.")]
head_params = [p for n, p in model.named_parameters() if p.requires_grad and n.startswith("fc.")]

param_groups = [{"params": head_params, "lr": HEAD_LR}]
if backbone_params:
    param_groups.append({"params": backbone_params, "lr": BACKBONE_LR})

optimizer = optim.Adam(param_groups, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2)

# 6. TRAIN / EVAL LOOP
def run_epoch(model, dataloader, criterion, optimizer=None):
    is_train = optimizer is not None
    model.train() if is_train else model.eval()

    running_loss, running_correct, total = 0.0, 0, 0
    torch.set_grad_enabled(is_train)

    for images, labels in dataloader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        if is_train:
            optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        if is_train:
            loss.backward()
            optimizer.step()
        preds = outputs.argmax(dim=1)
        running_loss += loss.item() * images.size(0)
        running_correct += (preds == labels).sum().item()
        total += images.size(0)

    return running_loss / total, running_correct / total


def train_model():
    best_val_loss = float("inf")
    best_state = copy.deepcopy(model.state_dict())
    epochs_no_improve = 0
    history = []

    for epoch in range(NUM_EPOCHS):
        start = time.time()
        train_loss, train_acc = run_epoch(model, train_dl, criterion, optimizer)
        val_loss, val_acc = run_epoch(model, val_dl, criterion, optimizer=None)
        scheduler.step(val_loss)

        elapsed = time.time() - start
        print(f"Epoch {epoch+1}/{NUM_EPOCHS} "
              f"[{elapsed:.0f}s] "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f}")

        history.append({
            "epoch": epoch + 1, "train_loss": train_loss, "train_acc": train_acc,
            "val_loss": val_loss, "val_acc": val_acc,
            "lr": optimizer.param_groups[0]["lr"]
        })

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = copy.deepcopy(model.state_dict())
            torch.save(best_state, CHECKPOINT_PATH)
            epochs_no_improve = 0
            print(f"  -> Saved new best checkpoint (val_loss={val_loss:.4f})")
        else:
            epochs_no_improve += 1
            print(f"  -> Không cải thiện ({epochs_no_improve}/{EARLY_STOP_PATIENCE})")
            if epochs_no_improve >= EARLY_STOP_PATIENCE:
                print(f"Early stopping tại epoch {epoch+1} (val_loss không cải thiện "
                      f"{EARLY_STOP_PATIENCE} epoch liên tiếp).")
                break

    model.load_state_dict(best_state)
    return model, history


# 7. TEST EVALUATION (đầy đủ chỉ số + confusion matrix + lưu Drive)
def evaluate_test(model, history=None):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []

    with torch.no_grad():
        for images, labels in test_dl:
            images = images.to(DEVICE)
            outputs = model(images)
            probs = torch.softmax(outputs, dim=1)[:, 1]
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_probs.extend(probs.cpu().numpy())

    class_names = list(train_ds.class_to_idx.keys())
    all_labels_np = np.array(all_labels)
    all_preds_np  = np.array(all_preds)
    all_probs_np  = np.array(all_probs)

    print("\n=== Test set results ===")
    print(classification_report(all_labels_np, all_preds_np, target_names=class_names))
    cm = confusion_matrix(all_labels_np, all_preds_np)
    print("Confusion matrix:\n", cm)

    metrics = {
        "accuracy": float((all_preds_np == all_labels_np).mean()),
        "balanced_accuracy": float(balanced_accuracy_score(all_labels_np, all_preds_np)),
        "f1_macro": float(f1_score(all_labels_np, all_preds_np, average="macro")),
        "f1_weighted": float(f1_score(all_labels_np, all_preds_np, average="weighted")),
        "precision_macro": float(precision_score(all_labels_np, all_preds_np, average="macro")),
        "recall_macro": float(recall_score(all_labels_np, all_preds_np, average="macro")),
        "mcc": float(matthews_corrcoef(all_labels_np, all_preds_np)),
        "roc_auc": float(roc_auc_score(all_labels_np, all_probs_np)),
        "pr_auc": float(average_precision_score(all_labels_np, all_probs_np)),
    }
    print("\n--- Chỉ số bổ sung ---")
    for k, v in metrics.items():
        print(f"{k:20s}: {v:.4f}")

    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(f"Confusion Matrix - {STRATEGY_NAME}")
    plt.tight_layout()
    cm_path_local = f"/content/{STRATEGY_NAME}_confusion_matrix.png"
    plt.savefig(cm_path_local, dpi=150)
    plt.show()
    shutil.copy2(cm_path_local, os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_confusion_matrix.png"))

    with open(os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_history.json"), "w") as f:
        json.dump({
            "strategy": STRATEGY_NAME,
            "confusion_matrix": cm.tolist(),
            "class_order": class_names,
            "test_metrics": metrics,
            "history": history if history is not None else [],
        }, f, indent=2)

    print(f"\nĐã lưu confusion matrix + chỉ số + lịch sử train vào: {RESULTS_DIR}")
    return metrics


# 8. RUN
if __name__ == "__main__":
    model, history = train_model()
    test_metrics = evaluate_test(model, history=history)

    drive_ckpt_path = os.path.join(CHECKPOINTS_DIR, f"{STRATEGY_NAME}_best.pth")
    shutil.copy2(CHECKPOINT_PATH, drive_ckpt_path)
    print(f"Đã lưu checkpoint tốt nhất vào Drive: {drive_ckpt_path}")
```

```python
# ==========================================================
# ResNet18 Deepfake Detection - v4 (Final Stability Edition)
# ==========================================================
import os
import re
import io
import copy
import time
import random
import json
import shutil
import warnings

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from PIL import Image
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, balanced_accuracy_score,
    matthews_corrcoef, f1_score, precision_score, recall_score
)

warnings.filterwarnings("ignore", category=UserWarning)

# ----------------------------
# 0. SEED - đặt NGAY sau import, TRƯỚC khi tạo model/dataset/dataloader
# ----------------------------
SEED = 42

def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)
print(f"Đã set seed = {SEED}")

# ==========================================================
# 1. CONFIG
# ==========================================================
BATCH_SIZE   = 32
NUM_WORKERS  = 2
NUM_EPOCHS   = 25

UNIFIED_LR   = 1e-4
WEIGHT_DECAY = 3e-4
DROPOUT      = 0.5
LABEL_SMOOTHING = 0.1
USE_HIDDEN_LAYER = False

EARLY_STOP_PATIENCE = 5

AUG_LEVEL = "super_heavy"
UNFREEZE_SCOPE = "layer3+4"

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/resnet18_deepfake_v4_final.pth"

STRATEGY_NAME = "v4_superheavy"   # >>> ĐÚNG cho file này, giữ nguyên
RESULTS_DIR      = "/content/drive/MyDrive/Deepfakeft/results"
CHECKPOINTS_DIR  = "/content/drive/MyDrive/Deepfakeft/checkpoints"
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINTS_DIR, exist_ok=True)

print(f"Device: {DEVICE}")

# ==========================================================
# 2. CHUẨN BỊ DATASET (tự phát hiện tar mới trên Drive)
# ==========================================================
TAR_FILE_DRIVE = "/content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar"
TAR_FILE_LOCAL = "/content/output_dataset_resplit.tar"
DATASET_ROOT   = "/content/output_dataset_resplit"
required_dirs  = ["train/real", "train/fake", "val/real", "val/fake", "test/real", "test/fake"]
SIGNATURE_FILE = os.path.join(DATASET_ROOT, ".source_signature")

def get_tar_signature(path):
    stat = os.stat(path)
    return f"{stat.st_size}_{int(stat.st_mtime)}"

def dataset_ready():
    dirs_ok = all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs)
    if not dirs_ok:
        return False
    if not os.path.isfile(SIGNATURE_FILE):
        return False
    with open(SIGNATURE_FILE) as f:
        cached_sig = f.read().strip()
    return cached_sig == get_tar_signature(TAR_FILE_DRIVE)

if not dataset_ready():
    if os.path.isdir(DATASET_ROOT):
        print("Phát hiện tar trên Drive đã đổi (hoặc dataset local không rõ nguồn) "
              "-> xoá dataset local cũ để giải nén lại...")
        get_ipython().system('rm -rf "$DATASET_ROOT"')

    src_size = os.path.getsize(TAR_FILE_DRIVE)
    need_copy = (not os.path.isfile(TAR_FILE_LOCAL)) or (os.path.getsize(TAR_FILE_LOCAL) != src_size)
    if need_copy:
        print(f"Đang copy tar về local ({src_size / 1e9:.2f} GB)...")
        get_ipython().system('cp "$TAR_FILE_DRIVE" "$TAR_FILE_LOCAL"')
        assert os.path.getsize(TAR_FILE_LOCAL) == src_size, "Copy tar bị dở dang/lỗi."
        print("Copy xong.")

    print("Đang giải nén...")
    get_ipython().system('tar -xf "$TAR_FILE_LOCAL" -C /content')
    assert all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs), \
        f"Giải nén xong nhưng thiếu thư mục con trong {DATASET_ROOT}."

    with open(SIGNATURE_FILE, "w") as f:
        f.write(get_tar_signature(TAR_FILE_DRIVE))
    print("Dataset sẵn sàng (đã đồng bộ với tar mới nhất trên Drive).")
else:
    print("Dataset đã sẵn sàng và khớp với tar hiện tại trên Drive.")

# ==========================================================
# 3. CHẨN ĐOÁN RÒ RỈ CLIP
# ==========================================================
def extract_clip_key(filename: str) -> str:
    name = os.path.splitext(filename)[0]
    name = re.sub(r'(_frame\d+|_f\d+|-\d{2,6}|_\d{3,6})$', '', name)
    return name

def check_clip_leakage(dataset_root, splits=("train", "val", "test"), classes=("real", "fake")):
    split_clip_sets = {}
    for split in splits:
        clip_set = set()
        for cls in classes:
            d = os.path.join(dataset_root, split, cls)
            if not os.path.isdir(d): continue
            for fname in os.listdir(d):
                clip_set.add(extract_clip_key(fname))
        split_clip_sets[split] = clip_set

    print("\n=== Chẩn đoán rò rỉ theo clip ===")
    any_overlap = False
    for i, s1 in enumerate(splits):
        for s2 in splits[i+1:]:
            overlap = split_clip_sets[s1] & split_clip_sets[s2]
            if overlap:
                any_overlap = True
                print(f"  CẢNH BÁO: {len(overlap)} clip trùng giữa {s1} và {s2}.")
            else:
                print(f"  OK: Không trùng clip giữa {s1} và {s2}.")
    if not any_overlap:
        print("  -> Dataset an toàn (theo tên file).")

try:
    check_clip_leakage(DATASET_ROOT)
except Exception as e:
    print(f"Bỏ qua chẩn đoán clip leakage (lỗi: {e}).")

# ==========================================================
# 4. TRANSFORMS (SUPER HEAVY AUGMENTATION)
# ==========================================================
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

class RandomJPEGCompression:
    def __init__(self, quality_range=(50, 95), p=0.5):
        self.quality_range = quality_range
        self.p = p

    def __call__(self, img: Image.Image) -> Image.Image:
        if random.random() > self.p:
            return img
        quality = random.randint(*self.quality_range)
        buffer = io.BytesIO()
        img.save(buffer, format="JPEG", quality=quality)
        buffer.seek(0)
        return Image.open(buffer).convert("RGB")

def build_train_transforms(level="super_heavy"):
    base = [
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1),
    ]
    if level == "super_heavy":
        extra = [
            transforms.RandomApply([transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0))], p=0.4),
            transforms.RandomAdjustSharpness(sharpness_factor=0.5, p=0.3),
            RandomJPEGCompression(quality_range=(50, 90), p=0.5),
            transforms.RandomRotation(degrees=10),
            transforms.RandomGrayscale(p=0.1),
        ]
    else:
        extra = []
    tail = [
        transforms.ToTensor(),
        transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
        transforms.RandomErasing(p=0.3, scale=(0.02, 0.15))
    ]
    return transforms.Compose(base + extra + tail)

train_transforms = build_train_transforms(AUG_LEVEL)
print(f"Augmentation level: {AUG_LEVEL}")

eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

# ==========================================================
# 5. DATASETS & DATALOADERS
# ==========================================================
train_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "train"), transform=train_transforms)
val_ds   = datasets.ImageFolder(os.path.join(DATASET_ROOT, "val"),   transform=eval_transforms)
test_ds  = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"),  transform=eval_transforms)

print("Class mapping:", train_ds.class_to_idx)
print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

# generator + worker_init_fn để shuffle=True cũng deterministic
g = torch.Generator()
g.manual_seed(SEED)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=NUM_WORKERS,
                       pin_memory=True, persistent_workers=(NUM_WORKERS > 0),
                       worker_init_fn=seed_worker, generator=g)
val_dl   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS,
                       pin_memory=True, persistent_workers=(NUM_WORKERS > 0))
test_dl  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS,
                       pin_memory=True, persistent_workers=(NUM_WORKERS > 0))

# ==========================================================
# 6. MODEL
# ==========================================================
def create_model(unfreeze_scope="layer3+4", dropout=0.5, use_hidden_layer=False):
    model = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
    for param in model.parameters():
        param.requires_grad = False
    if "layer3" in unfreeze_scope:
        for param in model.layer3.parameters():
            param.requires_grad = True
    if "layer4" in unfreeze_scope:
        for param in model.layer4.parameters():
            param.requires_grad = True

    in_features = model.fc.in_features
    if use_hidden_layer:
        model.fc = nn.Sequential(
            nn.Dropout(dropout), nn.Linear(in_features, in_features // 2),
            nn.ReLU(), nn.BatchNorm1d(in_features // 2),
            nn.Dropout(dropout), nn.Linear(in_features // 2, 2)
        )
    else:
        model.fc = nn.Sequential(nn.Dropout(dropout), nn.Linear(in_features, 2))
    return model.to(DEVICE)

def count_params(model):
    n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    n_total = sum(p.numel() for p in model.parameters())
    print(f"Trainable params: {n_trainable:,} / {n_total:,} ({100*n_trainable/n_total:.1f}%)")

# ==========================================================
# 7. LOSS & OPTIMIZER
# ==========================================================
criterion_counts = torch.bincount(torch.tensor(train_ds.targets))
class_weights = (criterion_counts.sum() / (len(criterion_counts) * criterion_counts)).float().to(DEVICE)
print(f"Class weights: {class_weights.tolist()}")

criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=LABEL_SMOOTHING)

# ==========================================================
# 8. TRAIN / EVAL LOOP
# ==========================================================
def run_epoch(model, dataloader, criterion, optimizer=None):
    is_train = optimizer is not None
    model.train() if is_train else model.eval()
    running_loss, running_correct, total = 0.0, 0, 0
    torch.set_grad_enabled(is_train)
    for images, labels in dataloader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        if is_train:
            optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        if is_train:
            loss.backward()
            optimizer.step()
        preds = outputs.argmax(dim=1)
        running_loss += loss.item() * images.size(0)
        running_correct += (preds == labels).sum().item()
        total += images.size(0)
    return running_loss / total, running_correct / total


def train_loop(model, optimizer, scheduler, num_epochs, ckpt_path, patience):
    best_val_acc = 0.0
    best_state = copy.deepcopy(model.state_dict())
    epochs_no_improve = 0
    history = []

    for epoch in range(num_epochs):
        start = time.time()
        train_loss, train_acc = run_epoch(model, train_dl, criterion, optimizer)
        val_loss, val_acc = run_epoch(model, val_dl, criterion, optimizer=None)

        if scheduler is not None:
            scheduler.step(val_loss)

        elapsed = time.time() - start
        print(f"Epoch {epoch+1}/{num_epochs} "
              f"[{elapsed:.0f}s] "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} | "
              f"LR: {optimizer.param_groups[0]['lr']:.2e}")

        history.append({
            "epoch": epoch + 1, "train_loss": train_loss, "train_acc": train_acc,
            "val_loss": val_loss, "val_acc": val_acc, "lr": optimizer.param_groups[0]['lr']
        })

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state = copy.deepcopy(model.state_dict())
            torch.save(best_state, ckpt_path)
            epochs_no_improve = 0
            print(f"  -> Saved best checkpoint (val_acc={val_acc:.4f})")
        else:
            epochs_no_improve += 1
            print(f"  -> Không cải thiện ({epochs_no_improve}/{patience})")
            if epochs_no_improve >= patience:
                print(f"Early stopping tại epoch {epoch+1}.")
                break

    model.load_state_dict(best_state)
    return model, best_val_acc, history


# ==========================================================
# 9. TEST EVALUATION (đầy đủ chỉ số + confusion matrix + lưu Drive)
# ==========================================================
def evaluate_test(model, history=None):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for images, labels in test_dl:
            images = images.to(DEVICE)
            outputs = model(images)
            probs = torch.softmax(outputs, dim=1)[:, 1]
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_probs.extend(probs.cpu().numpy())

    class_names = list(train_ds.class_to_idx.keys())
    all_labels_np = np.array(all_labels)
    all_preds_np  = np.array(all_preds)
    all_probs_np  = np.array(all_probs)

    print("\n" + "="*40)
    print(f"=== TEST SET RESULTS: {STRATEGY_NAME} ===")
    print("="*40)
    print(classification_report(all_labels_np, all_preds_np, target_names=class_names))

    cm = confusion_matrix(all_labels_np, all_preds_np)
    print("Confusion Matrix:\n", cm)

    metrics = {
        "accuracy": float((all_preds_np == all_labels_np).mean()),
        "balanced_accuracy": float(balanced_accuracy_score(all_labels_np, all_preds_np)),
        "f1_macro": float(f1_score(all_labels_np, all_preds_np, average="macro")),
        "f1_weighted": float(f1_score(all_labels_np, all_preds_np, average="weighted")),
        "precision_macro": float(precision_score(all_labels_np, all_preds_np, average="macro")),
        "recall_macro": float(recall_score(all_labels_np, all_preds_np, average="macro")),
        "mcc": float(matthews_corrcoef(all_labels_np, all_preds_np)),
        "roc_auc": float(roc_auc_score(all_labels_np, all_probs_np)),
        "pr_auc": float(average_precision_score(all_labels_np, all_probs_np)),
    }
    print("\n--- Chỉ số bổ sung ---")
    for k, v in metrics.items():
        print(f"{k:20s}: {v:.4f}")

    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(f"Confusion Matrix - {STRATEGY_NAME}")
    plt.tight_layout()
    cm_path_local = f"/content/{STRATEGY_NAME}_confusion_matrix.png"
    plt.savefig(cm_path_local, dpi=150)
    plt.show()
    shutil.copy2(cm_path_local, os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_confusion_matrix.png"))

    with open(os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_history.json"), "w") as f:
        json.dump({
            "strategy": STRATEGY_NAME,
            "confusion_matrix": cm.tolist(),
            "class_order": class_names,
            "test_metrics": metrics,
            "history": history if history is not None else [],
        }, f, indent=2)

    print(f"\nĐã lưu confusion matrix + chỉ số + lịch sử train vào: {RESULTS_DIR}")
    return metrics


# ==========================================================
# 10. CHẠY TRAINING
# ==========================================================
if __name__ == "__main__":
    print("\n" + "="*60)
    print("STRATEGY V4: Single-Stage + Super Heavy Aug + Track Val Acc")
    print("="*60)

    model = create_model(UNFREEZE_SCOPE, DROPOUT, USE_HIDDEN_LAYER)
    count_params(model)

    trainable_params = [p for p in model.parameters() if p.requires_grad]
    optimizer = optim.Adam(trainable_params, lr=UNIFIED_LR, weight_decay=WEIGHT_DECAY)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2)

    model, best_val_acc, history = train_loop(
        model, optimizer, scheduler,
        num_epochs=NUM_EPOCHS,
        ckpt_path=CHECKPOINT_PATH,
        patience=EARLY_STOP_PATIENCE
    )

    print(f"\nHoàn tất huấn luyện. Best Val Accuracy: {best_val_acc:.4f}")
    test_metrics = evaluate_test(model, history=history)

    drive_ckpt_path = os.path.join(CHECKPOINTS_DIR, f"{STRATEGY_NAME}_best.pth")
    shutil.copy2(CHECKPOINT_PATH, drive_ckpt_path)
    print(f"Đã lưu checkpoint tốt nhất vào Drive: {drive_ckpt_path}")
```

```python
# ==========================================================
# ResNet18 Deepfake Detection - V7 (The Pragmatic Sweet Spot)
# ==========================================================
from google.colab import drive
drive.mount('/content/drive')

import os, re, io, copy, time, random, json, shutil, warnings
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from PIL import Image
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, balanced_accuracy_score,
    matthews_corrcoef, f1_score, precision_score, recall_score
)

warnings.filterwarnings("ignore", category=UserWarning)

# ----------------------------
# 0. SEED - đặt NGAY sau import, TRƯỚC khi tạo model/dataset/dataloader
# ----------------------------
SEED = 42

def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)
print(f"Đã set seed = {SEED}")

# ==========================================================
# 1. CONFIG
# ==========================================================
BATCH_SIZE   = 32
NUM_WORKERS  = 2
NUM_EPOCHS   = 25

UNIFIED_LR   = 1e-4
WEIGHT_DECAY = 1e-3
DROPOUT      = 0.5
LABEL_SMOOTHING = 0.1
EARLY_STOP_PATIENCE = 5

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/resnet18_deepfake_v7_pragmatic.pth"

STRATEGY_NAME = "v7_pragmatic"   # >>> ĐÃ SỬA: trước bị nhầm thành "v4_superheavy"
RESULTS_DIR      = "/content/drive/MyDrive/Deepfakeft/results"
CHECKPOINTS_DIR  = "/content/drive/MyDrive/Deepfakeft/checkpoints"
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINTS_DIR, exist_ok=True)

print(f"Device: {DEVICE}")

# ----------------------------
# Chuẩn bị dataset
# ----------------------------
TAR_FILE_DRIVE = "/content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar"
TAR_FILE_LOCAL = "/content/output_dataset_resplit.tar"
DATASET_ROOT   = "/content/output_dataset_resplit"
required_dirs  = ["train/real", "train/fake", "val/real", "val/fake", "test/real", "test/fake"]
SIGNATURE_FILE = os.path.join(DATASET_ROOT, ".source_signature")

def get_tar_signature(path):
    stat = os.stat(path)
    return f"{stat.st_size}_{int(stat.st_mtime)}"

def dataset_ready():
    dirs_ok = all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs)
    if not dirs_ok:
        return False
    if not os.path.isfile(SIGNATURE_FILE):
        return False
    with open(SIGNATURE_FILE) as f:
        cached_sig = f.read().strip()
    return cached_sig == get_tar_signature(TAR_FILE_DRIVE)

if not dataset_ready():
    if os.path.isdir(DATASET_ROOT):
        print("Phát hiện tar trên Drive đã đổi (hoặc dataset local không rõ nguồn) "
              "-> xoá dataset local cũ để giải nén lại...")
        get_ipython().system('rm -rf "$DATASET_ROOT"')

    src_size = os.path.getsize(TAR_FILE_DRIVE)
    need_copy = (not os.path.isfile(TAR_FILE_LOCAL)) or (os.path.getsize(TAR_FILE_LOCAL) != src_size)
    if need_copy:
        print(f"Đang copy tar về local ({src_size / 1e9:.2f} GB)...")
        get_ipython().system('cp "$TAR_FILE_DRIVE" "$TAR_FILE_LOCAL"')
        assert os.path.getsize(TAR_FILE_LOCAL) == src_size, "Copy tar bị dở dang/lỗi."
        print("Copy xong.")

    print("Đang giải nén...")
    get_ipython().system('tar -xf "$TAR_FILE_LOCAL" -C /content')
    assert all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs), \
        f"Giải nén xong nhưng thiếu thư mục con trong {DATASET_ROOT}."

    with open(SIGNATURE_FILE, "w") as f:
        f.write(get_tar_signature(TAR_FILE_DRIVE))
    print("Dataset sẵn sàng (đã đồng bộ với tar mới nhất trên Drive).")
else:
    print("Dataset đã sẵn sàng và khớp với tar hiện tại trên Drive.")

# ==========================================================
# 3. TRANSFORMS (SUPER HEAVY AUG)
# ==========================================================
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

class RandomJPEGCompression:
    def __init__(self, quality_range=(50, 95), p=0.5):
        self.quality_range, self.p = quality_range, p
    def __call__(self, img):
        if random.random() > self.p: return img
        q = random.randint(*self.quality_range)
        buf = io.BytesIO()
        img.save(buf, format="JPEG", quality=q)
        buf.seek(0)
        return Image.open(buf).convert("RGB")

train_transforms = transforms.Compose([
    transforms.RandomResizedCrop(224, scale=(0.8, 1.0), ratio=(0.9, 1.1)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.RandomApply([transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0))], p=0.4),
    transforms.RandomAdjustSharpness(sharpness_factor=0.5, p=0.3),
    RandomJPEGCompression(quality_range=(50, 90), p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.RandomGrayscale(p=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    transforms.RandomErasing(p=0.2, scale=(0.02, 0.12))
])

eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

train_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "train"), transform=train_transforms)
val_ds   = datasets.ImageFolder(os.path.join(DATASET_ROOT, "val"),   transform=eval_transforms)
test_ds  = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"),  transform=eval_transforms)

print("Class mapping:", train_ds.class_to_idx)
print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

# generator + worker_init_fn để shuffle=True cũng deterministic
g = torch.Generator()
g.manual_seed(SEED)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=NUM_WORKERS,
                       pin_memory=True, worker_init_fn=seed_worker, generator=g)
val_dl   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_dl  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# ==========================================================
# 4. MODEL
# ==========================================================
def create_model():
    model = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
    for param in model.parameters():
        param.requires_grad = False
    for param in model.layer4.parameters():
        param.requires_grad = True
    in_features = model.fc.in_features
    model.fc = nn.Sequential(nn.Dropout(DROPOUT), nn.Linear(in_features, 2))
    return model.to(DEVICE)

# ==========================================================
# 5. LOSS & OPTIMIZER
# ==========================================================
criterion_counts = torch.bincount(torch.tensor(train_ds.targets))
class_weights = (criterion_counts.sum() / (len(criterion_counts) * criterion_counts)).float().to(DEVICE)
criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=LABEL_SMOOTHING)

# ==========================================================
# 6. TRAIN / EVAL LOOP
# ==========================================================
def run_epoch(model, dataloader, criterion, optimizer=None):
    is_train = optimizer is not None
    torch.set_grad_enabled(is_train)
    model.train() if is_train else model.eval()

    running_loss, running_correct, total = 0.0, 0, 0
    for images, labels in dataloader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        if is_train: optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        if is_train:
            loss.backward()
            optimizer.step()
        preds = outputs.argmax(dim=1)
        running_loss += loss.item() * images.size(0)
        running_correct += (preds == labels).sum().item()
        total += images.size(0)
    return running_loss / total, running_correct / total

def train_loop(model, optimizer, scheduler, num_epochs, ckpt_path, patience):
    best_val_acc, epochs_no_improve = 0.0, 0
    best_state = copy.deepcopy(model.state_dict())
    history = []

    for epoch in range(num_epochs):
        start = time.time()
        train_loss, train_acc = run_epoch(model, train_dl, criterion, optimizer)
        val_loss, val_acc = run_epoch(model, val_dl, criterion, optimizer=None)
        if scheduler: scheduler.step(val_loss)

        print(f"Epoch {epoch+1}/{num_epochs} [{time.time()-start:.0f}s] "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} | LR: {optimizer.param_groups[0]['lr']:.2e}")

        history.append({
            "epoch": epoch + 1, "train_loss": train_loss, "train_acc": train_acc,
            "val_loss": val_loss, "val_acc": val_acc, "lr": optimizer.param_groups[0]['lr']
        })

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state = copy.deepcopy(model.state_dict())
            torch.save(best_state, ckpt_path)
            epochs_no_improve = 0
            print(f"  -> Saved best (val_acc={val_acc:.4f})")
        else:
            epochs_no_improve += 1
            print(f"  -> No improve ({epochs_no_improve}/{patience})")
            if epochs_no_improve >= patience: break

    model.load_state_dict(best_state)
    return model, best_val_acc, history


# ==========================================================
# 7. TEST EVALUATION (đầy đủ chỉ số + confusion matrix + lưu Drive)
# ==========================================================
def evaluate_test(model, history=None):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []
    torch.set_grad_enabled(False)
    for images, labels in test_dl:
        images = images.to(DEVICE)
        outputs = model(images)
        probs = torch.softmax(outputs, dim=1)[:, 1]
        all_preds.extend(outputs.argmax(dim=1).cpu().numpy())
        all_labels.extend(labels.numpy())
        all_probs.extend(probs.cpu().numpy())

    class_names = list(train_ds.class_to_idx.keys())
    all_labels_np = np.array(all_labels)
    all_preds_np  = np.array(all_preds)
    all_probs_np  = np.array(all_probs)

    print("\n=== TEST SET FINAL RESULTS ===")
    print(classification_report(all_labels_np, all_preds_np, target_names=class_names))
    cm = confusion_matrix(all_labels_np, all_preds_np)
    print("Confusion Matrix:\n", cm)

    metrics = {
        "accuracy": float((all_preds_np == all_labels_np).mean()),
        "balanced_accuracy": float(balanced_accuracy_score(all_labels_np, all_preds_np)),
        "f1_macro": float(f1_score(all_labels_np, all_preds_np, average="macro")),
        "f1_weighted": float(f1_score(all_labels_np, all_preds_np, average="weighted")),
        "precision_macro": float(precision_score(all_labels_np, all_preds_np, average="macro")),
        "recall_macro": float(recall_score(all_labels_np, all_preds_np, average="macro")),
        "mcc": float(matthews_corrcoef(all_labels_np, all_preds_np)),
        "roc_auc": float(roc_auc_score(all_labels_np, all_probs_np)),
        "pr_auc": float(average_precision_score(all_labels_np, all_probs_np)),
    }
    print("\n--- Chỉ số bổ sung ---")
    for k, v in metrics.items():
        print(f"{k:20s}: {v:.4f}")

    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(f"Confusion Matrix - {STRATEGY_NAME}")
    plt.tight_layout()
    cm_path_local = f"/content/{STRATEGY_NAME}_confusion_matrix.png"
    plt.savefig(cm_path_local, dpi=150)
    plt.show()
    shutil.copy2(cm_path_local, os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_confusion_matrix.png"))

    with open(os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_history.json"), "w") as f:
        json.dump({
            "strategy": STRATEGY_NAME,
            "confusion_matrix": cm.tolist(),
            "class_order": class_names,
            "test_metrics": metrics,
            "history": history if history is not None else [],
        }, f, indent=2)

    print(f"\nĐã lưu confusion matrix + chỉ số + lịch sử train vào: {RESULTS_DIR}")
    return metrics


# ==========================================================
# 8. CHẠY TRAINING
# ==========================================================
if __name__ == "__main__":
    print("\n" + "="*60)
    print("STRATEGY V7: Pragmatic (Layer4 + SuperHeavy + No BN Fix + AdamW)")
    print("="*60)

    model = create_model()
    n_tr = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Trainable params: {n_tr:,}")

    optimizer = optim.AdamW(model.parameters(), lr=UNIFIED_LR, weight_decay=WEIGHT_DECAY)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2)

    model, best_val_acc, history = train_loop(
        model, optimizer, scheduler, NUM_EPOCHS, CHECKPOINT_PATH, EARLY_STOP_PATIENCE
    )
    print(f"\nDone. Best Val Acc: {best_val_acc:.4f}")
    test_metrics = evaluate_test(model, history=history)

    drive_ckpt_path = os.path.join(CHECKPOINTS_DIR, f"{STRATEGY_NAME}_best.pth")
    shutil.copy2(CHECKPOINT_PATH, drive_ckpt_path)
    print(f"Đã lưu checkpoint tốt nhất vào Drive: {drive_ckpt_path}")
```

```python
# ResNet18 Deepfake Detection - Siamese (BCE + Triplet + Batch Hard Miner)
# Dựa trên baseline ResNet18 (partial fine-tune, dataset đã identity-resplit),
# áp thêm hướng cải thiện Siamese từ model cũ: embedding_head (Triplet) +
# classifier_head (BCE), Joint Loss = BCE + 1.0*Triplet, AdamW + CosineAnnealingLR.

from google.colab import drive
drive.mount('/content/drive')
!pip install pytorch-metric-learning -q

import os
import time
import copy
import json
import random
import shutil
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, balanced_accuracy_score,
    matthews_corrcoef, f1_score, precision_score, recall_score
)
from pytorch_metric_learning import losses as pml_losses, miners
from pytorch_metric_learning.samplers import MPerClassSampler

# ----------------------------
# 0. SEED - đặt NGAY sau import, TRƯỚC khi tạo model/dataset/dataloader
# ----------------------------
SEED = 42

def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)          # multi-GPU (nếu có)
    torch.backends.cudnn.deterministic = True  # ép cuDNN chọn thuật toán deterministic
    torch.backends.cudnn.benchmark = False     # tắt auto-tune (auto-tune có thể phá tính deterministic)

set_seed(SEED)
print(f"Đã set seed = {SEED}")

# 1. CONFIG
BATCH_SIZE   = 32
M_PER_CLASS  = 16          # mỗi batch train luôn có 16 real + 16 fake -> đủ để Batch Hard Miner hoạt động
NUM_EPOCHS   = 15
WEIGHT_DECAY = 1e-4
NUM_WORKERS  = 2
DROPOUT      = 0.4         # theo bản Siamese cũ (0.4, nhẹ hơn 0.5 của baseline thường)
EMBEDDING_DIM  = 128
TRIPLET_MARGIN = 0.3
TRIPLET_WEIGHT = 1.0       # Total loss = BCE + 1.0*Triplet
FREEZE_MODE  = "partial"   # đã kiểm chứng tốt hơn "full"/"none" trên dataset sạch
BACKBONE_LR  = 1e-5
HEAD_LR      = 5e-5
EARLY_STOP_PATIENCE = 3
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/resnet18_siamese_best.pth"

STRATEGY_NAME = "siamese_triplet"
RESULTS_DIR      = "/content/drive/MyDrive/Deepfakeft/results"
CHECKPOINTS_DIR  = "/content/drive/MyDrive/Deepfakeft/checkpoints"
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINTS_DIR, exist_ok=True)

print(f"Device: {DEVICE}")

# ----------------------------
# Chuẩn bị dataset: copy tar về local rồi giải nén, có kiểm tra chữ ký
# (size + mtime) của tar trên Drive để tự động giải nén lại nếu resplit
# và ghi đè tar mới, tránh việc dataset_ready() báo nhầm "sẵn sàng".
# ----------------------------
TAR_FILE_DRIVE = "/content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar"
TAR_FILE_LOCAL = "/content/output_dataset_resplit.tar"
DATASET_ROOT   = "/content/output_dataset_resplit"
required_dirs  = ["train/real", "train/fake", "val/real", "val/fake", "test/real", "test/fake"]
SIGNATURE_FILE = os.path.join(DATASET_ROOT, ".source_signature")

def get_tar_signature(path):
    stat = os.stat(path)
    return f"{stat.st_size}_{int(stat.st_mtime)}"

def dataset_ready():
    dirs_ok = all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs)
    if not dirs_ok:
        return False
    if not os.path.isfile(SIGNATURE_FILE):
        return False
    with open(SIGNATURE_FILE) as f:
        cached_sig = f.read().strip()
    return cached_sig == get_tar_signature(TAR_FILE_DRIVE)

if not dataset_ready():
    if os.path.isdir(DATASET_ROOT):
        print("Phát hiện tar trên Drive đã đổi (hoặc dataset local không rõ nguồn) "
              "-> xoá dataset local cũ để giải nén lại...")
        get_ipython().system('rm -rf "$DATASET_ROOT"')

    src_size = os.path.getsize(TAR_FILE_DRIVE)
    need_copy = (not os.path.isfile(TAR_FILE_LOCAL)) or (os.path.getsize(TAR_FILE_LOCAL) != src_size)
    if need_copy:
        print(f"Đang copy tar về local ({src_size / 1e9:.2f} GB)...")
        get_ipython().system('cp "$TAR_FILE_DRIVE" "$TAR_FILE_LOCAL"')
        assert os.path.getsize(TAR_FILE_LOCAL) == src_size, "Copy tar bị dở dang/lỗi."
        print("Copy xong.")

    print("Đang giải nén...")
    get_ipython().system('tar -xf "$TAR_FILE_LOCAL" -C /content')
    assert all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs), \
        f"Giải nén xong nhưng thiếu thư mục con trong {DATASET_ROOT}."

    with open(SIGNATURE_FILE, "w") as f:
        f.write(get_tar_signature(TAR_FILE_DRIVE))
    print("Dataset sẵn sàng (đã đồng bộ với tar mới nhất trên Drive).")
else:
    print("Dataset đã sẵn sàng và khớp với tar hiện tại trên Drive.")

# 2. TRANSFORMS
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])
eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

# 3. DATASETS & DATALOADERS
train_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "train"), transform=train_transforms)
val_ds   = datasets.ImageFolder(os.path.join(DATASET_ROOT, "val"),   transform=eval_transforms)
test_ds  = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"),  transform=eval_transforms)

print("Class mapping:", train_ds.class_to_idx)
print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

train_sampler = MPerClassSampler(train_ds.targets, m=M_PER_CLASS, batch_size=BATCH_SIZE,
                                  length_before_new_iter=len(train_ds))

# generator riêng cho DataLoader để shuffle/worker cũng deterministic
g = torch.Generator()
g.manual_seed(SEED)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=train_sampler,
                       num_workers=NUM_WORKERS, pin_memory=True, drop_last=True,
                       worker_init_fn=seed_worker, generator=g)
val_dl   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False,
                       num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=(NUM_WORKERS > 0))
test_dl  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False,
                       num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=(NUM_WORKERS > 0))

# 4. MODEL
def create_siamese_model(embedding_dim=128, dropout=0.4, freeze_mode="partial"):
    backbone = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)

    if freeze_mode in ("full", "partial"):
        for param in backbone.parameters():
            param.requires_grad = False
    if freeze_mode == "partial":
        for param in backbone.layer4.parameters():
            param.requires_grad = True

    in_features = backbone.fc.in_features
    backbone.fc = nn.Identity()
    print(f"Input feature dim: {in_features}")

    class SiameseResNet18(nn.Module):
        def __init__(self):
            super().__init__()
            self.backbone = backbone
            self.embedding_head = nn.Sequential(
                nn.Linear(in_features, embedding_dim),
                nn.ReLU(),
                nn.Dropout(dropout)
            )
            self.classifier = nn.Sequential(
                nn.Dropout(dropout),
                nn.Linear(embedding_dim, 1)
            )

        def encode(self, x):
            feat = self.backbone(x)
            emb = self.embedding_head(feat)
            return F.normalize(emb, p=2, dim=1)

        def forward(self, x):
            emb = self.encode(x)
            logit = self.classifier(emb).squeeze(1)
            return emb, logit

    return SiameseResNet18().to(DEVICE)

model = create_siamese_model(EMBEDDING_DIM, DROPOUT, FREEZE_MODE)

n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
n_total = sum(p.numel() for p in model.parameters())
print(f"Trainable params: {n_trainable:,} / {n_total:,}")

# 5. LOSS / OPTIMIZER
class_counts = torch.bincount(torch.tensor(train_ds.targets))
class_weights = (class_counts.sum() / (len(class_counts) * class_counts)).float().to(DEVICE)
print(f"Class counts (theo class_to_idx {train_ds.class_to_idx}): {class_counts.tolist()}")
print(f"Class weights: {class_weights.tolist()}")

def weighted_bce(logits, labels):
    weights = class_weights[labels]
    return F.binary_cross_entropy_with_logits(logits, labels.float(), weight=weights)

miner = miners.BatchHardMiner()
triplet_loss_fn = pml_losses.TripletMarginLoss(margin=TRIPLET_MARGIN)

backbone_params = [p for n, p in model.named_parameters()
                    if p.requires_grad and n.startswith("backbone.")]
head_params = [p for n, p in model.named_parameters()
                if p.requires_grad and not n.startswith("backbone.")]
param_groups = [{"params": head_params, "lr": HEAD_LR}]
if backbone_params:
    param_groups.append({"params": backbone_params, "lr": BACKBONE_LR})

optimizer = optim.AdamW(param_groups, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS, eta_min=1e-6)

# 6. TRAIN / EVAL LOOP
def run_epoch(dataloader, is_train):
    model.train() if is_train else model.eval()
    total_bce, total_triplet, total_loss, correct, n = 0.0, 0.0, 0.0, 0, 0

    torch.set_grad_enabled(is_train)
    for images, labels in dataloader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        if is_train:
            optimizer.zero_grad()

        embeddings, logits = model(images)
        bce = weighted_bce(logits, labels)

        hard_pairs = miner(embeddings, labels)
        triplet = triplet_loss_fn(embeddings, labels, hard_pairs)

        loss = bce + TRIPLET_WEIGHT * triplet

        if is_train:
            loss.backward()
            optimizer.step()

        preds = (torch.sigmoid(logits) >= 0.5).long()
        correct += (preds == labels).sum().item()
        n += labels.size(0)
        total_bce += bce.item() * labels.size(0)
        total_triplet += triplet.item() * labels.size(0)
        total_loss += loss.item() * labels.size(0)

    return {"loss": total_loss / n, "bce": total_bce / n, "triplet": total_triplet / n, "acc": correct / n}


def train_model():
    best_val_loss = float("inf")
    best_state = copy.deepcopy(model.state_dict())
    epochs_no_improve = 0
    history = []

    for epoch in range(NUM_EPOCHS):
        start = time.time()
        tr = run_epoch(train_dl, is_train=True)
        va = run_epoch(val_dl, is_train=False)
        scheduler.step()
        current_lr = optimizer.param_groups[0]["lr"]
        elapsed = time.time() - start

        print(f"Epoch {epoch+1}/{NUM_EPOCHS} [{elapsed:.0f}s] "
              f"train: loss={tr['loss']:.4f} (bce={tr['bce']:.4f} triplet={tr['triplet']:.4f}) acc={tr['acc']:.4f} | "
              f"val: loss={va['loss']:.4f} (bce={va['bce']:.4f} triplet={va['triplet']:.4f}) acc={va['acc']:.4f} | "
              f"LR: {current_lr:.2e}")

        history.append({
            "epoch": epoch + 1,
            "train_loss": tr["loss"], "train_bce": tr["bce"], "train_triplet": tr["triplet"], "train_acc": tr["acc"],
            "val_loss": va["loss"], "val_bce": va["bce"], "val_triplet": va["triplet"], "val_acc": va["acc"],
            "lr": current_lr
        })

        if va["loss"] < best_val_loss:
            best_val_loss = va["loss"]
            best_state = copy.deepcopy(model.state_dict())
            torch.save(best_state, CHECKPOINT_PATH)
            epochs_no_improve = 0
            print(f"  -> Saved new best checkpoint (val_loss={va['loss']:.4f})")
        else:
            epochs_no_improve += 1
            print(f"  -> Không cải thiện ({epochs_no_improve}/{EARLY_STOP_PATIENCE})")
            if epochs_no_improve >= EARLY_STOP_PATIENCE:
                print(f"Early stopping tại epoch {epoch+1}.")
                break

    model.load_state_dict(best_state)
    return model, history


# 7. TEST EVALUATION
def evaluate_test(model, history=None):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for images, labels in test_dl:
            images = images.to(DEVICE)
            _, logits = model(images)
            probs = torch.sigmoid(logits)
            preds = (probs >= 0.5).long().cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_probs.extend(probs.cpu().numpy())

    class_names = list(train_ds.class_to_idx.keys())
    all_labels_np = np.array(all_labels)
    all_preds_np = np.array(all_preds)
    all_probs_np = np.array(all_probs)

    print("\n=== Test set results (dùng classifier head, không cần embedding lúc inference) ===")
    print(classification_report(all_labels_np, all_preds_np, target_names=class_names))
    cm = confusion_matrix(all_labels_np, all_preds_np)
    print("Confusion matrix:\n", cm)

    metrics = {
        "accuracy": float((all_preds_np == all_labels_np).mean()),
        "balanced_accuracy": float(balanced_accuracy_score(all_labels_np, all_preds_np)),
        "f1_macro": float(f1_score(all_labels_np, all_preds_np, average="macro")),
        "f1_weighted": float(f1_score(all_labels_np, all_preds_np, average="weighted")),
        "precision_macro": float(precision_score(all_labels_np, all_preds_np, average="macro")),
        "recall_macro": float(recall_score(all_labels_np, all_preds_np, average="macro")),
        "mcc": float(matthews_corrcoef(all_labels_np, all_preds_np)),
        "roc_auc": float(roc_auc_score(all_labels_np, all_probs_np)),
        "pr_auc": float(average_precision_score(all_labels_np, all_probs_np)),
    }
    print("\n--- Chỉ số bổ sung ---")
    for k, v in metrics.items():
        print(f"{k:20s}: {v:.4f}")

    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(f"Confusion Matrix - {STRATEGY_NAME}")
    plt.tight_layout()
    cm_path_local = f"/content/{STRATEGY_NAME}_confusion_matrix.png"
    plt.savefig(cm_path_local, dpi=150)
    plt.show()
    shutil.copy2(cm_path_local, os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_confusion_matrix.png"))

    with open(os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_history.json"), "w") as f:
        json.dump({
            "strategy": STRATEGY_NAME,
            "confusion_matrix": cm.tolist(),
            "class_order": class_names,
            "test_metrics": metrics,
            "history": history if history is not None else [],
        }, f, indent=2)

    print(f"\nĐã lưu confusion matrix + chỉ số + lịch sử train vào: {RESULTS_DIR}")
    return metrics


# 8. RUN
if __name__ == "__main__":
    model, history = train_model()
    test_metrics = evaluate_test(model, history=history)

    drive_ckpt_path = os.path.join(CHECKPOINTS_DIR, f"{STRATEGY_NAME}_best.pth")
    shutil.copy2(CHECKPOINT_PATH, drive_ckpt_path)
    print(f"Đã lưu checkpoint tốt nhất vào Drive: {drive_ckpt_path}")
```

```python
# ==========================================================
# Siamese Ablation: CHỈ BCE, TẮT TRIPLET LOSS (TRIPLET_WEIGHT=0)
# ==========================================================
from google.colab import drive
drive.mount('/content/drive')
!pip install pytorch-metric-learning -q

import os, time, copy, json, random, shutil
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, balanced_accuracy_score,
    matthews_corrcoef, f1_score, precision_score, recall_score
)
from pytorch_metric_learning import losses as pml_losses, miners
from pytorch_metric_learning.samplers import MPerClassSampler

# ----------------------------
# 0. SEED - đặt NGAY sau import, TRƯỚC khi tạo model/dataset/dataloader
# ----------------------------
SEED = 42

def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)
print(f"Đã set seed = {SEED}")

# 1. CONFIG
BATCH_SIZE   = 32
M_PER_CLASS  = 16
NUM_EPOCHS   = 15
WEIGHT_DECAY = 1e-4
NUM_WORKERS  = 2
DROPOUT      = 0.4
EMBEDDING_DIM  = 128
TRIPLET_MARGIN = 0.3
TRIPLET_WEIGHT = 0.0        # <<< ĐÃ TẮT TRIPLET, CHỈ CÒN BCE >>>
FREEZE_MODE  = "partial"
BACKBONE_LR  = 1e-5
HEAD_LR      = 5e-5
EARLY_STOP_PATIENCE = 3
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/resnet18_siamese_ablation_bce_only.pth"

STRATEGY_NAME = "siamese_ablation_bce_only"
RESULTS_DIR      = "/content/drive/MyDrive/Deepfakeft/results"
CHECKPOINTS_DIR  = "/content/drive/MyDrive/Deepfakeft/checkpoints"
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINTS_DIR, exist_ok=True)

print(f"Device: {DEVICE}")

# ----------------------------
# Chuẩn bị dataset (đồng bộ cách check tar như các script khác)
# ----------------------------
TAR_FILE_DRIVE = "/content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar"
TAR_FILE_LOCAL = "/content/output_dataset_resplit.tar"
DATASET_ROOT   = "/content/output_dataset_resplit"
required_dirs  = ["train/real", "train/fake", "val/real", "val/fake", "test/real", "test/fake"]
SIGNATURE_FILE = os.path.join(DATASET_ROOT, ".source_signature")

def get_tar_signature(path):
    stat = os.stat(path)
    return f"{stat.st_size}_{int(stat.st_mtime)}"

def dataset_ready():
    dirs_ok = all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs)
    if not dirs_ok:
        return False
    if not os.path.isfile(SIGNATURE_FILE):
        return False
    with open(SIGNATURE_FILE) as f:
        cached_sig = f.read().strip()
    return cached_sig == get_tar_signature(TAR_FILE_DRIVE)

if not dataset_ready():
    if os.path.isdir(DATASET_ROOT):
        print("Phát hiện tar trên Drive đã đổi (hoặc dataset local không rõ nguồn) "
              "-> xoá dataset local cũ để giải nén lại...")
        get_ipython().system('rm -rf "$DATASET_ROOT"')

    src_size = os.path.getsize(TAR_FILE_DRIVE)
    need_copy = (not os.path.isfile(TAR_FILE_LOCAL)) or (os.path.getsize(TAR_FILE_LOCAL) != src_size)
    if need_copy:
        print(f"Đang copy tar về local ({src_size / 1e9:.2f} GB)...")
        get_ipython().system('cp "$TAR_FILE_DRIVE" "$TAR_FILE_LOCAL"')
        assert os.path.getsize(TAR_FILE_LOCAL) == src_size, "Copy tar bị dở dang/lỗi."
        print("Copy xong.")

    print("Đang giải nén...")
    get_ipython().system('tar -xf "$TAR_FILE_LOCAL" -C /content')
    assert all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs), \
        f"Giải nén xong nhưng thiếu thư mục con trong {DATASET_ROOT}."

    with open(SIGNATURE_FILE, "w") as f:
        f.write(get_tar_signature(TAR_FILE_DRIVE))
    print("Dataset sẵn sàng (đã đồng bộ với tar mới nhất trên Drive).")
else:
    print("Dataset đã sẵn sàng và khớp với tar hiện tại trên Drive.")

# 2. TRANSFORMS
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])
eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

train_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "train"), transform=train_transforms)
val_ds   = datasets.ImageFolder(os.path.join(DATASET_ROOT, "val"),   transform=eval_transforms)
test_ds  = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"),  transform=eval_transforms)

print("Class mapping:", train_ds.class_to_idx)
print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

train_sampler = MPerClassSampler(train_ds.targets, m=M_PER_CLASS, batch_size=BATCH_SIZE,
                                  length_before_new_iter=len(train_ds))

# generator + worker_init_fn để sampler cũng deterministic
g = torch.Generator()
g.manual_seed(SEED)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=train_sampler,
                       num_workers=NUM_WORKERS, pin_memory=True, drop_last=True,
                       worker_init_fn=seed_worker, generator=g)
val_dl   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False,
                       num_workers=NUM_WORKERS, pin_memory=True)
test_dl  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False,
                       num_workers=NUM_WORKERS, pin_memory=True)

def create_siamese_model(embedding_dim=128, dropout=0.4, freeze_mode="partial"):
    backbone = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
    if freeze_mode in ("full", "partial"):
        for param in backbone.parameters():
            param.requires_grad = False
    if freeze_mode == "partial":
        for param in backbone.layer4.parameters():
            param.requires_grad = True
    in_features = backbone.fc.in_features
    backbone.fc = nn.Identity()

    class SiameseResNet18(nn.Module):
        def __init__(self):
            super().__init__()
            self.backbone = backbone
            self.embedding_head = nn.Sequential(
                nn.Linear(in_features, embedding_dim), nn.ReLU(), nn.Dropout(dropout)
            )
            self.classifier = nn.Sequential(nn.Dropout(dropout), nn.Linear(embedding_dim, 1))
        def encode(self, x):
            return F.normalize(self.embedding_head(self.backbone(x)), p=2, dim=1)
        def forward(self, x):
            emb = self.encode(x)
            return emb, self.classifier(emb).squeeze(1)

    return SiameseResNet18().to(DEVICE)

model = create_siamese_model(EMBEDDING_DIM, DROPOUT, FREEZE_MODE)
n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
n_total = sum(p.numel() for p in model.parameters())
print(f"Trainable params: {n_trainable:,} / {n_total:,}")

class_counts = torch.bincount(torch.tensor(train_ds.targets))
class_weights = (class_counts.sum() / (len(class_counts) * class_counts)).float().to(DEVICE)
print(f"Class weights: {class_weights.tolist()}")

def weighted_bce(logits, labels):
    weights = class_weights[labels]
    return F.binary_cross_entropy_with_logits(logits, labels.float(), weight=weights)

miner = miners.BatchHardMiner()
triplet_loss_fn = pml_losses.TripletMarginLoss(margin=TRIPLET_MARGIN)

backbone_params = [p for n, p in model.named_parameters() if p.requires_grad and n.startswith("backbone.")]
head_params = [p for n, p in model.named_parameters() if p.requires_grad and not n.startswith("backbone.")]
param_groups = [{"params": head_params, "lr": HEAD_LR}]
if backbone_params:
    param_groups.append({"params": backbone_params, "lr": BACKBONE_LR})

optimizer = optim.AdamW(param_groups, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS, eta_min=1e-6)

def run_epoch(dataloader, is_train):
    model.train() if is_train else model.eval()
    total_bce, total_triplet, total_loss, correct, n = 0.0, 0.0, 0.0, 0, 0
    torch.set_grad_enabled(is_train)
    for images, labels in dataloader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        if is_train: optimizer.zero_grad()
        embeddings, logits = model(images)
        bce = weighted_bce(logits, labels)
        # Vẫn TÍNH triplet để theo dõi/so sánh, nhưng KHÔNG ảnh hưởng gradient (TRIPLET_WEIGHT=0)
        hard_pairs = miner(embeddings, labels)
        triplet = triplet_loss_fn(embeddings, labels, hard_pairs)
        loss = bce + TRIPLET_WEIGHT * triplet
        if is_train:
            loss.backward()
            optimizer.step()
        preds = (torch.sigmoid(logits) >= 0.5).long()
        correct += (preds == labels).sum().item()
        n += labels.size(0)
        total_bce += bce.item() * labels.size(0)
        total_triplet += triplet.item() * labels.size(0)
        total_loss += loss.item() * labels.size(0)
    return {"loss": total_loss/n, "bce": total_bce/n, "triplet": total_triplet/n, "acc": correct/n}


def train_model():
    best_val_loss = float("inf")
    best_state = copy.deepcopy(model.state_dict())
    epochs_no_improve = 0
    history = []

    for epoch in range(NUM_EPOCHS):
        start = time.time()
        tr = run_epoch(train_dl, is_train=True)
        va = run_epoch(val_dl, is_train=False)
        scheduler.step()
        current_lr = optimizer.param_groups[0]["lr"]
        elapsed = time.time() - start

        print(f"Epoch {epoch+1}/{NUM_EPOCHS} [{elapsed:.0f}s] "
              f"train: loss={tr['loss']:.4f} acc={tr['acc']:.4f} | "
              f"val: loss={va['loss']:.4f} acc={va['acc']:.4f} | "
              f"LR: {current_lr:.2e}")

        history.append({
            "epoch": epoch + 1,
            "train_loss": tr["loss"], "train_bce": tr["bce"], "train_triplet": tr["triplet"], "train_acc": tr["acc"],
            "val_loss": va["loss"], "val_bce": va["bce"], "val_triplet": va["triplet"], "val_acc": va["acc"],
            "lr": current_lr
        })

        if va["loss"] < best_val_loss:
            best_val_loss = va["loss"]
            best_state = copy.deepcopy(model.state_dict())
            torch.save(best_state, CHECKPOINT_PATH)
            epochs_no_improve = 0
            print(f"  -> Saved new best checkpoint (val_loss={va['loss']:.4f})")
        else:
            epochs_no_improve += 1
            print(f"  -> Không cải thiện ({epochs_no_improve}/{EARLY_STOP_PATIENCE})")
            if epochs_no_improve >= EARLY_STOP_PATIENCE:
                print(f"Early stopping tại epoch {epoch+1}.")
                break

    model.load_state_dict(best_state)
    return model, history


def evaluate_test(model, history=None):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for images, labels in test_dl:
            images = images.to(DEVICE)
            _, logits = model(images)
            probs = torch.sigmoid(logits)
            preds = (probs >= 0.5).long().cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_probs.extend(probs.cpu().numpy())

    class_names = list(train_ds.class_to_idx.keys())
    all_labels_np = np.array(all_labels)
    all_preds_np = np.array(all_preds)
    all_probs_np = np.array(all_probs)

    print("\n=== Test set results (Ablation: BCE only, no Triplet) ===")
    print(classification_report(all_labels_np, all_preds_np, target_names=class_names))
    cm = confusion_matrix(all_labels_np, all_preds_np)
    print("Confusion matrix:\n", cm)

    metrics = {
        "accuracy": float((all_preds_np == all_labels_np).mean()),
        "balanced_accuracy": float(balanced_accuracy_score(all_labels_np, all_preds_np)),
        "f1_macro": float(f1_score(all_labels_np, all_preds_np, average="macro")),
        "f1_weighted": float(f1_score(all_labels_np, all_preds_np, average="weighted")),
        "precision_macro": float(precision_score(all_labels_np, all_preds_np, average="macro")),
        "recall_macro": float(recall_score(all_labels_np, all_preds_np, average="macro")),
        "mcc": float(matthews_corrcoef(all_labels_np, all_preds_np)),
        "roc_auc": float(roc_auc_score(all_labels_np, all_probs_np)),
        "pr_auc": float(average_precision_score(all_labels_np, all_probs_np)),
    }
    print("\n--- Chỉ số bổ sung ---")
    for k, v in metrics.items():
        print(f"{k:20s}: {v:.4f}")

    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(f"Confusion Matrix - {STRATEGY_NAME}")
    plt.tight_layout()
    cm_path_local = f"/content/{STRATEGY_NAME}_confusion_matrix.png"
    plt.savefig(cm_path_local, dpi=150)
    plt.show()
    shutil.copy2(cm_path_local, os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_confusion_matrix.png"))

    with open(os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_history.json"), "w") as f:
        json.dump({
            "strategy": STRATEGY_NAME,
            "confusion_matrix": cm.tolist(),
            "class_order": class_names,
            "test_metrics": metrics,
            "history": history if history is not None else [],
        }, f, indent=2)

    print(f"\nĐã lưu confusion matrix + chỉ số + lịch sử train vào: {RESULTS_DIR}")
    return metrics


if __name__ == "__main__":
    model, history = train_model()
    test_metrics = evaluate_test(model, history=history)

    drive_ckpt_path = os.path.join(CHECKPOINTS_DIR, f"{STRATEGY_NAME}_best.pth")
    shutil.copy2(CHECKPOINT_PATH, drive_ckpt_path)
    print(f"Đã lưu checkpoint tốt nhất vào Drive: {drive_ckpt_path}")
```

```python
# ResNet18 Deepfake Detection - Siamese (BCE + Triplet + Batch Hard Miner)
# Dựa trên baseline ResNet18 (partial fine-tune, dataset đã identity-resplit),
# áp thêm hướng cải thiện Siamese từ model cũ: embedding_head (Triplet) +
# classifier_head (BCE), Joint Loss = BCE + 0.5*Triplet, AdamW + CosineAnnealingLR.

from google.colab import drive
drive.mount('/content/drive')
!pip install pytorch-metric-learning -q

import os
import time
import copy
import json
import random
import shutil
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
from torch.utils.data import DataLoader
from torchvision import datasets, models, transforms
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, balanced_accuracy_score,
    matthews_corrcoef, f1_score, precision_score, recall_score
)
from pytorch_metric_learning import losses as pml_losses, miners
from pytorch_metric_learning.samplers import MPerClassSampler

# ----------------------------
# 0. SEED - đặt NGAY sau import, TRƯỚC khi tạo model/dataset/dataloader
# ----------------------------
SEED = 42

def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)
print(f"Đã set seed = {SEED}")

# 1. CONFIG
BATCH_SIZE   = 32
M_PER_CLASS  = 16          # mỗi batch train luôn có 16 real + 16 fake -> đủ để Batch Hard Miner hoạt động
NUM_EPOCHS   = 15
WEIGHT_DECAY = 1e-4
NUM_WORKERS  = 2
DROPOUT      = 0.4         # theo bản Siamese cũ (0.4, nhẹ hơn 0.5 của baseline thường)
EMBEDDING_DIM  = 128
TRIPLET_MARGIN = 0.3
TRIPLET_WEIGHT = 0.5       # Total loss = BCE + 0.5*Triplet
FREEZE_MODE  = "partial"   # đã kiểm chứng tốt hơn "full"/"none" trên dataset sạch
BACKBONE_LR  = 1e-5
HEAD_LR      = 5e-5
EARLY_STOP_PATIENCE = 3
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/resnet18_siamese_w05_best.pth"

STRATEGY_NAME = "siamese_triplet_w0.5"
RESULTS_DIR      = "/content/drive/MyDrive/Deepfakeft/results"
CHECKPOINTS_DIR  = "/content/drive/MyDrive/Deepfakeft/checkpoints"
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(CHECKPOINTS_DIR, exist_ok=True)

print(f"Device: {DEVICE}")

# ----------------------------
# Chuẩn bị dataset: copy tar về local rồi giải nén, có kiểm tra chữ ký
# (size + mtime) của tar trên Drive để tự động giải nén lại nếu resplit
# và ghi đè tar mới, tránh việc dataset_ready() báo nhầm "sẵn sàng".
# ----------------------------
TAR_FILE_DRIVE = "/content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar"
TAR_FILE_LOCAL = "/content/output_dataset_resplit.tar"
DATASET_ROOT   = "/content/output_dataset_resplit"
required_dirs  = ["train/real", "train/fake", "val/real", "val/fake", "test/real", "test/fake"]
SIGNATURE_FILE = os.path.join(DATASET_ROOT, ".source_signature")

def get_tar_signature(path):
    stat = os.stat(path)
    return f"{stat.st_size}_{int(stat.st_mtime)}"

def dataset_ready():
    dirs_ok = all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs)
    if not dirs_ok:
        return False
    if not os.path.isfile(SIGNATURE_FILE):
        return False
    with open(SIGNATURE_FILE) as f:
        cached_sig = f.read().strip()
    return cached_sig == get_tar_signature(TAR_FILE_DRIVE)

if not dataset_ready():
    if os.path.isdir(DATASET_ROOT):
        print("Phát hiện tar trên Drive đã đổi (hoặc dataset local không rõ nguồn) "
              "-> xoá dataset local cũ để giải nén lại...")
        get_ipython().system('rm -rf "$DATASET_ROOT"')

    src_size = os.path.getsize(TAR_FILE_DRIVE)
    need_copy = (not os.path.isfile(TAR_FILE_LOCAL)) or (os.path.getsize(TAR_FILE_LOCAL) != src_size)
    if need_copy:
        print(f"Đang copy tar về local ({src_size / 1e9:.2f} GB)...")
        get_ipython().system('cp "$TAR_FILE_DRIVE" "$TAR_FILE_LOCAL"')
        assert os.path.getsize(TAR_FILE_LOCAL) == src_size, "Copy tar bị dở dang/lỗi."
        print("Copy xong.")

    print("Đang giải nén...")
    get_ipython().system('tar -xf "$TAR_FILE_LOCAL" -C /content')
    assert all(os.path.isdir(os.path.join(DATASET_ROOT, d)) for d in required_dirs), \
        f"Giải nén xong nhưng thiếu thư mục con trong {DATASET_ROOT}."

    with open(SIGNATURE_FILE, "w") as f:
        f.write(get_tar_signature(TAR_FILE_DRIVE))
    print("Dataset sẵn sàng (đã đồng bộ với tar mới nhất trên Drive).")
else:
    print("Dataset đã sẵn sàng và khớp với tar hiện tại trên Drive.")

# 2. TRANSFORMS
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])
eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

# 3. DATASETS & DATALOADERS
train_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "train"), transform=train_transforms)
val_ds   = datasets.ImageFolder(os.path.join(DATASET_ROOT, "val"),   transform=eval_transforms)
test_ds  = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"),  transform=eval_transforms)

print("Class mapping:", train_ds.class_to_idx)
print(f"Train: {len(train_ds)} | Val: {len(val_ds)} | Test: {len(test_ds)}")

train_sampler = MPerClassSampler(train_ds.targets, m=M_PER_CLASS, batch_size=BATCH_SIZE,
                                  length_before_new_iter=len(train_ds))

# generator + worker_init_fn để sampler cũng deterministic
g = torch.Generator()
g.manual_seed(SEED)

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

train_dl = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=train_sampler,
                       num_workers=NUM_WORKERS, pin_memory=True, drop_last=True,
                       worker_init_fn=seed_worker, generator=g)
val_dl   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False,
                       num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=(NUM_WORKERS > 0))
test_dl  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False,
                       num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=(NUM_WORKERS > 0))

# 4. MODEL
def create_siamese_model(embedding_dim=128, dropout=0.4, freeze_mode="partial"):
    backbone = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)

    if freeze_mode in ("full", "partial"):
        for param in backbone.parameters():
            param.requires_grad = False
    if freeze_mode == "partial":
        for param in backbone.layer4.parameters():
            param.requires_grad = True

    in_features = backbone.fc.in_features
    backbone.fc = nn.Identity()
    print(f"Input feature dim: {in_features}")

    class SiameseResNet18(nn.Module):
        def __init__(self):
            super().__init__()
            self.backbone = backbone
            self.embedding_head = nn.Sequential(
                nn.Linear(in_features, embedding_dim),
                nn.ReLU(),
                nn.Dropout(dropout)
            )
            self.classifier = nn.Sequential(
                nn.Dropout(dropout),
                nn.Linear(embedding_dim, 1)
            )

        def encode(self, x):
            feat = self.backbone(x)
            emb = self.embedding_head(feat)
            return F.normalize(emb, p=2, dim=1)

        def forward(self, x):
            emb = self.encode(x)
            logit = self.classifier(emb).squeeze(1)
            return emb, logit

    return SiameseResNet18().to(DEVICE)

model = create_siamese_model(EMBEDDING_DIM, DROPOUT, FREEZE_MODE)

n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
n_total = sum(p.numel() for p in model.parameters())
print(f"Trainable params: {n_trainable:,} / {n_total:,}")

# 5. LOSS / OPTIMIZER
class_counts = torch.bincount(torch.tensor(train_ds.targets))
class_weights = (class_counts.sum() / (len(class_counts) * class_counts)).float().to(DEVICE)
print(f"Class counts (theo class_to_idx {train_ds.class_to_idx}): {class_counts.tolist()}")
print(f"Class weights: {class_weights.tolist()}")

def weighted_bce(logits, labels):
    weights = class_weights[labels]
    return F.binary_cross_entropy_with_logits(logits, labels.float(), weight=weights)

miner = miners.BatchHardMiner()
triplet_loss_fn = pml_losses.TripletMarginLoss(margin=TRIPLET_MARGIN)

backbone_params = [p for n, p in model.named_parameters()
                    if p.requires_grad and n.startswith("backbone.")]
head_params = [p for n, p in model.named_parameters()
                if p.requires_grad and not n.startswith("backbone.")]
param_groups = [{"params": head_params, "lr": HEAD_LR}]
if backbone_params:
    param_groups.append({"params": backbone_params, "lr": BACKBONE_LR})

optimizer = optim.AdamW(param_groups, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS, eta_min=1e-6)

# 6. TRAIN / EVAL LOOP
def run_epoch(dataloader, is_train):
    model.train() if is_train else model.eval()
    total_bce, total_triplet, total_loss, correct, n = 0.0, 0.0, 0.0, 0, 0

    torch.set_grad_enabled(is_train)
    for images, labels in dataloader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        if is_train:
            optimizer.zero_grad()

        embeddings, logits = model(images)
        bce = weighted_bce(logits, labels)

        hard_pairs = miner(embeddings, labels)
        triplet = triplet_loss_fn(embeddings, labels, hard_pairs)

        loss = bce + TRIPLET_WEIGHT * triplet

        if is_train:
            loss.backward()
            optimizer.step()

        preds = (torch.sigmoid(logits) >= 0.5).long()
        correct += (preds == labels).sum().item()
        n += labels.size(0)
        total_bce += bce.item() * labels.size(0)
        total_triplet += triplet.item() * labels.size(0)
        total_loss += loss.item() * labels.size(0)

    return {"loss": total_loss / n, "bce": total_bce / n, "triplet": total_triplet / n, "acc": correct / n}


def train_model():
    best_val_loss = float("inf")
    best_state = copy.deepcopy(model.state_dict())
    epochs_no_improve = 0
    history = []

    for epoch in range(NUM_EPOCHS):
        start = time.time()
        tr = run_epoch(train_dl, is_train=True)
        va = run_epoch(val_dl, is_train=False)
        scheduler.step()
        current_lr = optimizer.param_groups[0]["lr"]
        elapsed = time.time() - start

        print(f"Epoch {epoch+1}/{NUM_EPOCHS} [{elapsed:.0f}s] "
              f"train: loss={tr['loss']:.4f} (bce={tr['bce']:.4f} triplet={tr['triplet']:.4f}) acc={tr['acc']:.4f} | "
              f"val: loss={va['loss']:.4f} (bce={va['bce']:.4f} triplet={va['triplet']:.4f}) acc={va['acc']:.4f} | "
              f"LR: {current_lr:.2e}")

        history.append({
            "epoch": epoch + 1,
            "train_loss": tr["loss"], "train_bce": tr["bce"], "train_triplet": tr["triplet"], "train_acc": tr["acc"],
            "val_loss": va["loss"], "val_bce": va["bce"], "val_triplet": va["triplet"], "val_acc": va["acc"],
            "lr": current_lr
        })

        if va["loss"] < best_val_loss:
            best_val_loss = va["loss"]
            best_state = copy.deepcopy(model.state_dict())
            torch.save(best_state, CHECKPOINT_PATH)
            epochs_no_improve = 0
            print(f"  -> Saved new best checkpoint (val_loss={va['loss']:.4f})")
        else:
            epochs_no_improve += 1
            print(f"  -> Không cải thiện ({epochs_no_improve}/{EARLY_STOP_PATIENCE})")
            if epochs_no_improve >= EARLY_STOP_PATIENCE:
                print(f"Early stopping tại epoch {epoch+1}.")
                break

    model.load_state_dict(best_state)
    return model, history


# 7. TEST EVALUATION
def evaluate_test(model, history=None):
    model.eval()
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for images, labels in test_dl:
            images = images.to(DEVICE)
            _, logits = model(images)
            probs = torch.sigmoid(logits)
            preds = (probs >= 0.5).long().cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_probs.extend(probs.cpu().numpy())

    class_names = list(train_ds.class_to_idx.keys())
    all_labels_np = np.array(all_labels)
    all_preds_np = np.array(all_preds)
    all_probs_np = np.array(all_probs)

    print("\n=== Test set results (dùng classifier head, không cần embedding lúc inference) ===")
    print(classification_report(all_labels_np, all_preds_np, target_names=class_names))
    cm = confusion_matrix(all_labels_np, all_preds_np)
    print("Confusion matrix:\n", cm)

    metrics = {
        "accuracy": float((all_preds_np == all_labels_np).mean()),
        "balanced_accuracy": float(balanced_accuracy_score(all_labels_np, all_preds_np)),
        "f1_macro": float(f1_score(all_labels_np, all_preds_np, average="macro")),
        "f1_weighted": float(f1_score(all_labels_np, all_preds_np, average="weighted")),
        "precision_macro": float(precision_score(all_labels_np, all_preds_np, average="macro")),
        "recall_macro": float(recall_score(all_labels_np, all_preds_np, average="macro")),
        "mcc": float(matthews_corrcoef(all_labels_np, all_preds_np)),
        "roc_auc": float(roc_auc_score(all_labels_np, all_probs_np)),
        "pr_auc": float(average_precision_score(all_labels_np, all_probs_np)),
    }
    print("\n--- Chỉ số bổ sung ---")
    for k, v in metrics.items():
        print(f"{k:20s}: {v:.4f}")

    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(f"Confusion Matrix - {STRATEGY_NAME}")
    plt.tight_layout()
    cm_path_local = f"/content/{STRATEGY_NAME}_confusion_matrix.png"
    plt.savefig(cm_path_local, dpi=150)
    plt.show()
    shutil.copy2(cm_path_local, os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_confusion_matrix.png"))

    with open(os.path.join(RESULTS_DIR, f"{STRATEGY_NAME}_history.json"), "w") as f:
        json.dump({
            "strategy": STRATEGY_NAME,
            "confusion_matrix": cm.tolist(),
            "class_order": class_names,
            "test_metrics": metrics,
            "history": history if history is not None else [],
        }, f, indent=2)

    print(f"\nĐã lưu confusion matrix + chỉ số + lịch sử train vào: {RESULTS_DIR}")
    return metrics


# 8. RUN
if __name__ == "__main__":
    model, history = train_model()
    test_metrics = evaluate_test(model, history=history)

    drive_ckpt_path = os.path.join(CHECKPOINTS_DIR, f"{STRATEGY_NAME}_best.pth")
    shutil.copy2(CHECKPOINT_PATH, drive_ckpt_path)
    print(f"Đã lưu checkpoint tốt nhất vào Drive: {drive_ckpt_path}")
```

```python
import os
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, precision_recall_curve, auc, confusion_matrix
import torch
import torch.nn as nn
from torchvision import models, transforms, datasets
from torch.utils.data import DataLoader
from IPython.display import display

# --- 0. GIẢI NÉN DỮ LIỆU TỰ ĐỘNG (FIX LỖI) ---
if not os.path.exists('/content/output_dataset_resplit/test'):
    print("0. Đang giải nén dữ liệu hình ảnh từ Drive (sẽ mất khoảng 1-2 phút)...")
    subprocess.run('cp /content/drive/MyDrive/Deepfakeft/output_dataset_resplit.tar /content/', shell=True)
    subprocess.run('tar -xf /content/output_dataset_resplit.tar -C /content', shell=True)
    print("-> Đã giải nén xong ảnh!")

# --- 1. CÀI ĐẶT MÔI TRƯỜNG & ĐƯỜNG DẪN ---
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DATASET_ROOT = "/content/output_dataset_resplit"
WEIGHTS_PATH = "/content/drive/MyDrive/Deepfakeft/checkpoints/v7_pragmatic_best.pth"

print("1. Đang chuẩn bị dữ liệu test...")
IMAGENET_MEAN, IMAGENET_STD = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]
eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)), transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])
test_ds = datasets.ImageFolder(os.path.join(DATASET_ROOT, "test"), transform=eval_transforms)
test_dl = DataLoader(test_ds, batch_size=32, shuffle=False, num_workers=2)

print("2. Đang xây dựng lại mô hình (V7) và nạp trọng số...")
model = models.resnet18(weights=None)
in_features = model.fc.in_features
model.fc = nn.Sequential(nn.Dropout(0.5), nn.Linear(in_features, 2))
model.load_state_dict(torch.load(WEIGHTS_PATH, map_location=DEVICE))
model = model.to(DEVICE)
model.eval()

print("3. Đang quét hình ảnh để lấy xác suất (chờ thêm khoảng 1 phút)...")
all_labels, all_probs = [], []
with torch.no_grad():
    for images, labels in test_dl:
        images = images.to(DEVICE)
        outputs = model(images)
        probs = torch.softmax(outputs, dim=1)[:, 1] # Lấy xác suất của class 1 (Real)
        all_labels.extend(labels.numpy())
        all_probs.extend(probs.cpu().numpy())

y_true = np.array(all_labels)
y_scores = np.array(all_probs)

print("4. Đang vẽ biểu đồ và tính toán ngưỡng tối ưu...\n")
# --- TÍNH TOÁN ROC & PR ---
fpr, tpr, roc_thresholds = roc_curve(y_true, y_scores)
roc_auc = auc(fpr, tpr)
precision, recall, pr_thresholds = precision_recall_curve(y_true, y_scores)
pr_auc = auc(recall, precision)

# Tìm ngưỡng tối ưu
best_idx_roc = np.argmax(tpr - fpr)
best_thresh_roc = roc_thresholds[best_idx_roc]
f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)
best_idx_pr = np.argmax(f1_scores)
best_thresh_pr = pr_thresholds[best_idx_pr]

# --- VẼ BIỂU ĐỒ ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(fpr, tpr, color='blue', label=f'ROC curve (AUC = {roc_auc:.3f})')
ax1.plot([0, 1], [0, 1], color='gray', linestyle='--')
ax1.scatter(fpr[best_idx_roc], tpr[best_idx_roc], color='red', s=80, label=f'Optimal (Youden J): Thresh={best_thresh_roc:.3f}')
ax1.set_xlabel('False Positive Rate (FPR)'); ax1.set_ylabel('True Positive Rate (TPR)')
ax1.set_title('Receiver Operating Characteristic (ROC)'); ax1.legend(loc="lower right"); ax1.grid(alpha=0.3)

ax2.plot(recall, precision, color='green', label=f'PR curve (AUC = {pr_auc:.3f})')
ax2.scatter(recall[best_idx_pr], precision[best_idx_pr], color='red', s=80, label=f'Optimal (Max F1): Thresh={best_thresh_pr:.3f}')
ax2.set_xlabel('Recall'); ax2.set_ylabel('Precision')
ax2.set_title('Precision-Recall (PR) Curve'); ax2.legend(loc="lower left"); ax2.grid(alpha=0.3)
plt.tight_layout(); plt.show()

# --- XUẤT BẢNG SO SÁNH ---
def eval_thresh(thresh):
    y_pred = (y_scores >= thresh).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    acc = (tp + tn) / len(y_true)
    prec = tp / (tp + fp) if (tp + fp) > 0 else 0
    rec = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (prec * rec) / (prec + rec) if (prec + rec) > 0 else 0
    return [thresh, acc, prec, rec, f1, fp, fn]

results = [eval_thresh(0.5), eval_thresh(best_thresh_roc), eval_thresh(best_thresh_pr)]
df_res = pd.DataFrame(results, columns=['Threshold', 'Accuracy', 'Precision', 'Recall', 'F1-Score', 'Báo nhầm Real (FP)', 'Bỏ lọt Real (FN)'])
df_res['Phương pháp'] = ['Mặc định (0.5)', 'Tối ưu ROC (Youden)', 'Tối ưu PR (Max F1)']
df_res.set_index('Phương pháp', inplace=True)

print("=== BẢNG SO SÁNH HIỆU SUẤT TRƯỚC VÀ SAU KHI SỬA THRESHOLD ===")
display(df_res.round(4))
```
